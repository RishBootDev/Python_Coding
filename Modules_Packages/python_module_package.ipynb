# Python Notes: Module and Packages

These notes explain how Python imports modules, functions, variables, and classes. Run the code cells from top to bottom in Jupyter Notebook.


## 1. What is a module?

A **module** is a Python file (`.py`) containing code such as functions, variables, and classes. Importing lets you reuse that code in another Python file or notebook instead of writing it again.

Python also includes standard-library modules, such as `math`, `random`, `statistics`, and `datetime`. You can install and import third-party packages too.

## 2. `import module_name`

Use `import` to import a module. Access its members with dot notation: `module_name.member_name`.

For example, `math.sqrt(25)` calls the `sqrt` function from the `math` module.

````python
import math

print(math.sqrt(25))
print(math.pi)
print(math.factorial(5))
````

**Output:**
```text
5.0
3.141592653589793
120
```

**Key point:** `import math` imports the module name `math`. You normally access its functions and constants as `math.sqrt`, `math.pi`, and so on.

## 3. `from module_name import member_name`

Use this form when you want to import specific members from a module. You can then use those names directly without the module prefix.

````python
from math import sqrt, pi, factorial

print(sqrt(36))
print(pi)
print(factorial(4))
````

**Output:**
```text
6.0
3.141592653589793
24
```

Here, `sqrt`, `pi`, and `factorial` are available directly. You do not need to write `math.sqrt(...)`.

## 4. Difference between `import` and `from ... import ...`

| Syntax | How you use a member | Example |
|---|---|---|
| `import math` | Use the module prefix | `math.sqrt(49)` |
| `from math import sqrt` | Use the imported member directly | `sqrt(49)` |

Both forms can be correct. `import module` makes the source of a name explicit; `from module import name` can be convenient when you need a few specific names frequently.

## 5. Import with an alias: `as`

The `as` keyword gives an imported module or member a different local name. This is useful for shorter names or conventional aliases.

````python
import math as m
from math import factorial as fact

print(m.sqrt(81))
print(fact(5))
````

**Output:**
```text
9.0
120
```

Common examples include `import numpy as np` and `import pandas as pd` (after installing those packages).

## 6. Import multiple members

You can import multiple named members in one statement by separating them with commas.

````python
from math import ceil, floor, pow

print(ceil(4.2))   # Round upward to an integer
print(floor(4.8))  # Round downward to an integer
print(pow(2, 3))   # 2 raised to the power 3
````

**Output:**
```text
5
4
8.0
```

## 7. `from module import *` (wildcard import)

A wildcard import attempts to bring many public names from a module into the current namespace:

```python
from math import *
print(sqrt(16))
print(pi)
```

This may work, but it is generally **not recommended** in regular code because it is unclear where names came from and names can collide. Prefer `import math` or explicitly import the names you need.

## 9. Importing a class

Classes can be imported just like functions. For example, the standard-library `datetime` module provides the `date` class.

````python
from datetime import date, datetime

today = date.today()
now = datetime.now()

print(today)
print(now.year)
````

The date and time values depend on when you run the cell. `from datetime import date` imports the `date` class from the `datetime` module; the module and class happen to have related names.

## 10. Importing from a package

A **package** is a way to organize related modules in directories. A simplified project might look like this:

```text
my_project/
    main.py
    utilities/
        __init__.py
        numbers.py
```

If `numbers.py` defines `is_even`, code in `main.py` may use:

```python
from utilities.numbers import is_even
```

or:

```python
import utilities.numbers
print(utilities.numbers.is_even(8))
```

The exact import depends on your project structure and where Python is run from. `__init__.py` is commonly used to mark or initialize a regular package, although modern Python also supports namespace packages without it.

## 11. Standard-library modules you can try

These modules are included with Python, so you usually do not need to install them.

````python
import random
import statistics
from collections import Counter

numbers = [2, 4, 4, 6, 8]

print(statistics.mean(numbers))
print(Counter(numbers))
print(random.randint(1, 10))  # Random integer; output can vary
````

## 13. Common errors and how to fix them

### `ModuleNotFoundError`
Python cannot find the requested module. Check the spelling, whether the package is installed, and whether your notebook uses the expected Python environment.

### `ImportError: cannot import name ...`
The module exists, but the requested name may not exist in it, may be spelled incorrectly, or may not be available in that version.

### Naming your file after a standard module
Avoid naming your own files `math.py`, `random.py`, `datetime.py`, or names of packages you use. Your file can shadow the real module and cause confusing import errors. If you rename such a file, remove its related `__pycache__` entry if needed and restart the kernel.

### Circular imports
If module A imports module B while module B imports module A, one may be only partially initialized when the other uses it. Move shared code into a third module or restructure the imports.

## 14. What happens when Python imports a module?

Python locates the module using its import system, loads and executes its module-level code the first time it is imported in that running process, and makes its names available. Python commonly caches loaded modules in `sys.modules`, so a later import usually reuses the loaded module rather than executing it from scratch.

This is why top-level statements in a module should be used thoughtfully: they run as part of importing it.

## 15. The `if __name__ == "__main__"` pattern

When a Python file is run directly, Python sets its `__name__` to `"__main__"`. When the file is imported as a module, its `__name__` is normally the module name. This pattern lets you keep test or demo code from running automatically when another file imports the module.

Example for `calculator.py`:

````python
# calculator.py
def add(a, b):
    return a + b

if __name__ == "__main__":
    print(add(2, 3))
````

Run `python calculator.py` to print `5`. If another file imports `calculator`, the function is defined, but the guarded `print` does not run automatically.
